```python
import cv2
import numpy as np
import os
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Dense,Conv2D,MaxPool2D,Flatten
```

```python
path=r'test111/airplane'
x=[]
y=[]
for h in os.listdir(path):
    if h.endswith('tif'):
        path2=os.path.join(path,h)
        img=cv2.imread(path2)
        img2=cv2.resize(img,(128,128))
        x.append(img2)
        y.append(0)
```

```python
path3=r'test111/river'
for t in os.listdir(path3):
    if t.endswith('tif'):
        path2=os.path.join(path3,t)
        img=cv2.imread(path2)
        img2=cv2.resize(img,(128,128))
        x.append(img2)
        y.append(1)
```

```python
path4=r'test111/harbor'
for r in os.listdir(path4):
    if r.endswith('tif'):
        path2=os.path.join(path4,r)
        img=cv2.imread(path2)
        img2=cv2.resize(img,(128,128))
        x.append(img2)
        y.append(2)
```

```python
path5=r'test111/forest'
for u in os.listdir(path5):
    if u.endswith('tif'):
        path2=os.path.join(path5,u)
        img=cv2.imread(path2)
        img2=cv2.resize(img,(128,128))
        x.append(img2)
        y.append(3)
```

```python
path6=r'test111/beach'
for o in os.listdir(path6):
    if o.endswith('tif'):
        path2=os.path.join(path6,o)
        img=cv2.imread(path2)
        img2=cv2.resize(img,(128,128))
        x.append(img2)
        y.append(4)
```

```python
x=np.array(x)
y=np.array(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape,y_train.shape,x_test.shape,y_test.shape)
```

```python
model=Sequential()
model.add(Conv2D(32,(3,3),padding='same',activation='relu',input_shape=(128,128,3)))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),padding='same',activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),padding='same',activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),padding='same',activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),padding='same',activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Conv2D(32,(3,3),padding='same',activation='relu'))
model.add(MaxPool2D(2,2))
model.add(Flatten())
model.add(Dense(5,activation='softmax'))
model.summary()
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=50,validation_split=0.1)
```

```python
model.evaluate(x_test,y_test)
```
